### 1.8 Outlier Exploration (Mencari Nilai Outlier)

Melakukan eksplorasi terhadap kemungkinan keberadaan outliers
pada atribut numerik menggunakan Boxplot dan metode
Interquartile Range (IQR).

IQR digunakan sebagai pendekatan awal untuk mengidentifikasi
observasi yang berada di luar rentang Q1 - 1.5 IQR
hingga Q3 + 1.5 IQR.

catatan: hasil analisis tidak langsung digunakan untuk menghapus data.
Setiap fitur akan dianalisis berdasarkan karakteristik dan
konteks bisnisnya.

```python
# Visualizing numerical features using boxplots

for column in numerical_columns:

    plt.figure(figsize=(10, 3))

    sns.boxplot(x=df[column])

    plt.title(f"Boxplot of {column}")
    plt.tight_layout()
    plt.show()
```

```python
# Detecting potential outliers using IQR

outlier_results = []

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ).sum()

    outlier_results.append({
        "Feature": column,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_count,
        "Outlier Percentage": outlier_count / len(df) * 100
    })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary)
```

Hasil identifikasi outlier menggunakan metode Interquartile Range (IQR) menunjukkan bahwa TransactionAmount memiliki 113 nilai yang berada di luar batas IQR atau sekitar 4,50% dari nilai valid pada atribut tersebut. Hal ini mengindikasikan adanya sejumlah transaksi dengan nominal yang relatif jauh lebih tinggi dibandingkan rentang mayoritas data.

Sementara itu CustomerAge, TransactionDuration, dan AccountBalance tidak menunjukkan observasi yang berada di luar batas IQR berdasarkan perhitungan yang digunakan.

Di sisi lain, LoginAttempts menghasilkan 120 observasi yang terdeteksi sebagai outlier. Namun, hasil ini perlu diinterpretasikan secara khusus karena Q1 dan Q3 pada atribut tersebut sama-sama bernilai 1, sehingga nilai IQR menjadi nol.

Secara keseluruhan, hasil identifikasi menunjukkan bahwa TransactionAmount merupakan atribut yang paling perlu diperiksa lebih lanjut terkait nilai ekstrem.

Namun, observasi ekstrem tidak langsung dihapus karena dapat merepresentasikan aktivitas transaksi yang sah.

### 1.10 Date Features Inspection

Memeriksa atribut tanggal transaksi dan tanggal transaksi
sebelumnya.

Tahap ini mencakup konversi sementara ke format datetime,
pemeriksaan rentang tanggal, serta identifikasi kemungkinan
inkonsistensi kronologis. Pemeriksaan ini penting dilakukan sebelum nantinya membentuk **fitur berbasis waktu,** seperti **recency** atau **transaction interval.**

```python
# Creating temporary datetime representations

date_columns = [
    "TransactionDate",
    "PreviousTransactionDate"
]

date_inspection = df[date_columns].copy()

for column in date_columns:

    date_inspection[column] = pd.to_datetime(
        date_inspection[column],
        errors="coerce"
    )

    print(f"\n{column}")
    print("Minimum:", date_inspection[column].min())
    print("Maximum:", date_inspection[column].max())
    print("Invalid dates:", date_inspection[column].isna().sum())

# Checking chronological inconsistencies
date_inconsistency = (
    date_inspection["PreviousTransactionDate"] >
    date_inspection["TransactionDate"]
)

print(
    "\nPrevious date later than current transaction:",
    date_inconsistency.sum()
)

display(
    df.loc[
        date_inconsistency,
        date_columns + ["AccountID"]
    ].head(20)
)
```

Hasil pemeriksaan tanggal menunjukkan bahwa TransactionDate memiliki rentang waktu mulai dari 2 Januari 2023 hingga 1 Januari 2024.

Sementara itu, PreviousTransactionDate memiliki rentang waktu yang sangat berbeda, yaitu seluruh nilai valid berada pada tanggal 4 November 2024 dengan rentang waktu sekitar enam menit.

Selain itu, ditemukan sebanyak 2.485 records dengan PreviousTransactionDate yang lebih baru daripada TransactionDate.

### 2.5 Resolving Duplicate TransactionID (Menangani Duplikat data pada Atribut TransactionID)

Melakukan pemeriksaan dan penyelesaian TransactionID yang
muncul lebih dari satu kali setelah full-row duplicates dihapus.

Berdasarkan hasil EDA, ditemukan tiga TransactionID yang
masing-masing muncul dalam dua records.

Setiap pasangan records diperiksa berdasarkan nilai atributnya.

Strategi penyelesaian:
1. Jika atribut yang terisi memiliki nilai yang sama dan
   atribut lainnya kosong, informasi akan digabungkan
   menggunakan nilai non-missing.
2. Jika terdapat dua nilai non-missing yang berbeda pada
   atribut yang sama, records akan ditandai sebagai konflik.
3. Records dengan konflik tidak akan dihapus atau digabungkan
   secara otomatis sebelum dilakukan pemeriksaan lebih lanjut.
4. Records dengan TransactionID kosong tetap dipertahankan
   karena belum tentu merupakan transaksi duplikat.

Tujuan proses ini adalah mempertahankan informasi sebanyak
mungkin tanpa menciptakan atau mengubah data secara sembarangan.

```python
# Resolving duplicate TransactionID records

resolved_records = []
conflict_records = []
merge_audit = []

# Separate records with valid TransactionID
valid_id_data = df_stage2[
    df_stage2["TransactionID"].notna()
]

# Process each TransactionID group
for transaction_id, group in valid_id_data.groupby(
    "TransactionID",
    sort=False
):

    # Keep records with unique TransactionID
    if len(group) == 1:

        resolved_records.append(
            group.iloc[0].to_dict()
        )

    else:

        # Check conflicting non-missing values
        conflicting_columns = [
            column
            for column in df_stage2.columns
            if group[column].dropna().nunique() > 1
        ]

        # Store unresolved conflicts
        if conflicting_columns:

            conflict_records.append(group)

        else:

            # Merge complementary records
            merged_row = group.iloc[0].copy()

            for column in df_stage2.columns:

                available_values = group[column].dropna()

                if not available_values.empty:

                    merged_row[column] = available_values.iloc[0]

            resolved_records.append(
                merged_row.to_dict()
            )

            merge_audit.append({
                "TransactionID": transaction_id,
                "Original Records": len(group),
                "Resolution": "Complementary records merged"
            })


# Preserve records with missing TransactionID
missing_id_records = df_stage2[
    df_stage2["TransactionID"].isna()
]

resolved_records.extend(
    missing_id_records.to_dict("records")
)


# Create cleaned transaction dataset
df_transactions = pd.DataFrame(
    resolved_records,
    columns=df_stage2.columns
)


# Create conflict dataset
if conflict_records:

    df_transaction_conflicts = pd.concat(
        conflict_records,
        ignore_index=True
    )

else:

    df_transaction_conflicts = pd.DataFrame(
        columns=df_stage2.columns
    )


# Display merge audit
merge_audit_df = pd.DataFrame(merge_audit)

print("Merged TransactionID Groups:", len(merge_audit_df))
print("Unresolved Conflict Groups:", len(conflict_records))
print("Remaining Duplicate Non-null TransactionID:",
      df_transactions.loc[
          df_transactions["TransactionID"].notna(),
          "TransactionID"
      ].duplicated().sum())

display(merge_audit_df)

display(df_transaction_conflicts)
```

Hasil pemeriksaan menemukan tiga kelompok TransactionID yang muncul lebih dari satu kali setelah penghapusan full-row duplicates. Ketiga kelompok tersebut memiliki informasi yang saling melengkapi dan tidak menunjukkan konflik nilai antaratribut.

Oleh karena itu, records berhasil digabungkan dengan mempertahankan informasi yang tersedia pada masing-masing atribut.

Proses ini menghasilkan dataset tanpa duplikasi TransactionID non-missing yang belum terselesaikan, dengan tetap mempertahankan records yang memiliki TransactionID kosong.

### 2.7 Temporal Data Validation (Validasi Data Kronologis)

Melakukan validasi terhadap konsistensi waktu pada atribut tanggal.
Berdasarkan hasil EDA, ditemukan bahwa sebagian besar
PreviousTransactionDate justru memiliki tanggal yang lebih
baru daripada TransactionDate. Kondisi tersebut tidak sesuai dengan makna transaksi sebelumnya.

Oleh karena itu, atribut PreviousTransactionDate akan
dipertahankan untuk keperluan audit, tetapi tidak digunakan
dalam pembentukan fitur temporal maupun clustering.

Sebaliknya, TransactionDate tetap dipertahankan karena
dapat digunakan untuk membentuk fitur aktivitas transaksi
seperti jumlah bulan aktif dan rentang waktu transaksi.

```python
# Validating chronological consistency

valid_date_mask = (
    df_customer_transactions["TransactionDate"].notna() &
    df_customer_transactions["PreviousTransactionDate"].notna()
)

previous_date_after_current = (
    df_customer_transactions.loc[valid_date_mask,
                                 "PreviousTransactionDate"] >
    df_customer_transactions.loc[valid_date_mask,
                                 "TransactionDate"]
)

invalid_date_count = previous_date_after_current.sum()

valid_date_count = valid_date_mask.sum()

print("Records with both valid dates:", valid_date_count)

print("Previous date later than current transaction:",
      invalid_date_count)

print("Chronological inconsistency percentage:",
      round(invalid_date_count / valid_date_count * 100, 2)
      if valid_date_count > 0 else 0)


# Displaying examples of inconsistent dates

temporal_issues = df_customer_transactions.loc[
    valid_date_mask
].loc[
    previous_date_after_current
][[
    "AccountID",
    "TransactionDate",
    "PreviousTransactionDate"
]]

display(temporal_issues.head(10))
```

Hasil validasi menunjukkan bahwa terdapat 2.442 records yang memiliki kedua atribut tanggal secara lengkap. Seluruh pasangan tanggal tersebut menunjukkan bahwa PreviousTransactionDate justru lebih baru daripada TransactionDate.

Kondisi ini mengindikasikan adanya inkonsistensi temporal yang serius pada dataset.Oleh karena itu, PreviousTransactionDate tidak digunakan dalam pembentukan fitur temporal maupun clustering, sedangkan TransactionDate tetap dipertahankan untuk mengidentifikasi aktivitas transaksi berdasarkan waktu.

## 2.8 Customer Profile Consistency Analysis (Memeriksa Konsistensi Atribut Customer)

Memeriksa konsistensi atribut CustomerAge dan CustomerOccupation
pada masing-masing AccountID. Berdasarkan hasil EDA, kedua atribut menunjukkan variasi nilai
yang cukup tinggi dalam satu AccountID. Hal ini perlu diperiksa karena usia dan pekerjaan secara umum
merupakan karakteristik profil pelanggan yang relatif tetap
dalam periode pengamatan yang pendek. Apabila satu AccountID memiliki beberapa nilai usia atau
pekerjaan yang berbeda, informasi tersebut belum dapat
dianggap sebagai profil pelanggan yang konsisten.

Oleh karena itu, CustomerAge dan CustomerOccupation tidak
akan dimasukkan sebagai fitur utama customer segmentation
pada eksperimen awal. Hasil pemeriksaan tetap disimpan sebagai dokumentasi
kualitas data.

```python
# Analyzing customer profile consistency

profile_consistency = (
    df_customer_transactions
    .groupby("AccountID")
    .agg(
        CustomerAge_Unique_Count=(
            "CustomerAge", "nunique"
        ),

        CustomerOccupation_Unique_Count=(
            "CustomerOccupation", "nunique"
        )
    )
)

# Counting inconsistent customer profiles

inconsistent_age = (
    profile_consistency["CustomerAge_Unique_Count"] > 1
).sum()

inconsistent_occupation = (
    profile_consistency["CustomerOccupation_Unique_Count"] > 1
).sum()

total_accounts = len(profile_consistency)

profile_summary = pd.DataFrame({
    "Feature": [
        "CustomerAge",
        "CustomerOccupation"
    ],

    "Accounts with Multiple Values": [
        inconsistent_age,
        inconsistent_occupation
    ],

    "Percentage": [
        inconsistent_age / total_accounts * 100,
        inconsistent_occupation / total_accounts * 100
    ]
})

display(profile_summary)
```

Hasil analisis menunjukkan bahwa sebagian besar AccountID memiliki lebih dari satu nilai CustomerAge maupun CustomerOccupation.

### 2.9 Numerical Feature Engineering: Transaction Preparation

Mempersiapkan variabel pendukung untuk pembentukan fitur
numerik pada akun pelanggan.

Proses ini mencakup:

1. Membentuk atribut bulan transaksi berdasarkan TransactionDate.
2. Memisahkan nominal transaksi Debit dan Credit.
3. Mempertahankan nilai numerik asli tanpa melakukan scaling.
4. Menggunakan informasi yang tersedia tanpa mengimputasi
   nominal transaksi yang kosong secara langsung.

Pemisahan nominal Debit dan Credit dilakukan karena kedua
jenis transaksi memiliki karakteristik finansial yang berbeda. Atribut PreviousTransactionDate tidak digunakan karena
hasil validasi sebelumnya menunjukkan ketidaksesuaian kronologis.

```python
# Preparing transaction-level feature variables

df_customer_transactions["TransactionMonth"] = (
    df_customer_transactions["TransactionDate"]
    .dt.to_period("M")
    .astype("string")
)

# Separating Debit and Credit transaction amounts

df_customer_transactions["Debit_Amount"] = (
    df_customer_transactions["TransactionAmount"]
    .where(
        df_customer_transactions["TransactionType"] == "Debit"
    )
)

df_customer_transactions["Credit_Amount"] = (
    df_customer_transactions["TransactionAmount"]
    .where(
        df_customer_transactions["TransactionType"] == "Credit"
    )
)

print("Transaction preparation completed.")

display(
    df_customer_transactions[
        [
            "AccountID",
            "TransactionAmount",
            "TransactionType",
            "Debit_Amount",
            "Credit_Amount",
            "TransactionDate",
            "TransactionMonth"
        ]
    ].head(10)
)
```

Proses persiapan fitur berhasil dilakukan dengan membentuk atribut **bulan transaksi** serta memisahkan **nominal transaksi** berdasarkan jenis Debit dan Credit.

Pemisahan ini memungkinkan analisis karakteristik transaksi keluar dan transaksi masuk secara terpisah.

Fitur TransactionMonth juga disiapkan untuk mengukur aktivitas transaksi setiap akun berdasarkan periode waktu.

### 2.10 Numerical Aggregation at Customer Level

Melakukan agregasi fitur numerik berdasarkan AccountID.

Setiap AccountID akan direpresentasikan melalui sejumlah
karakteristik yang merangkum seluruh records transaksi
yang dimilikinya.

Fitur yang dibentuk meliputi:

A. Transaction Activity
- Transaction Record Count
- Number of observed TransactionID
- Transaction activity span
- Number of active months

B. Transaction Amount
- Observed total transaction amount
- Average transaction amount
- Median transaction amount
- Standard deviation of transaction amount

C. Financial Characteristics
- Average account balance
- Median account balance
- Standard deviation of account balance
- Observed Debit amount
- Observed Credit amount

D. Transaction Behavior
- Average transaction duration
- Median transaction duration
- Average login attempts
- Maximum login attempts

E. Transaction Diversity
- Number of unique merchants
- Number of unique locations
- Number of unique devices

Agregasi numerik menggunakan fungsi statistik yang
mengabaikan missing values pada masing-masing fitur.

Khusus jumlah nominal, digunakan sum(min_count=1) untuk
menghindari hasil nol ketika seluruh nilai nominal pada
suatu kelompok tidak tersedia.

```python
# Aggregating numerical features by AccountID

customer_numeric_features = (
    df_customer_transactions
    .groupby("AccountID")
    .agg(
        # Transaction activity
        Transaction_Record_Count=(
            "AccountID", "size"
        ),

        TransactionID_Observed_Count=(
            "TransactionID", "count"
        ),

        TransactionID_Missing_Count=(
            "TransactionID",
            lambda x: x.isna().sum()
        ),

        Amount_Missing_Count=(
            "TransactionAmount",
            lambda x: x.isna().sum()
        ),

        # Transaction amount
        Observed_Transaction_Amount_Sum=(
            "TransactionAmount",
            lambda x: x.sum(min_count=1)
        ),

        Average_Transaction_Amount=(
            "TransactionAmount", "mean"
        ),

        Median_Transaction_Amount=(
            "TransactionAmount", "median"
        ),

        Std_Transaction_Amount=(
            "TransactionAmount", "std"
        ),

        # Debit and Credit amounts
        Debit_Amount_Sum=(
            "Debit_Amount",
            lambda x: x.sum(min_count=1)
        ),

        Credit_Amount_Sum=(
            "Credit_Amount",
            lambda x: x.sum(min_count=1)
        ),

        # Account balance
        Average_Account_Balance=(
            "AccountBalance", "mean"
        ),

        Median_Account_Balance=(
            "AccountBalance", "median"
        ),

        Std_Account_Balance=(
            "AccountBalance", "std"
        ),

        # Transaction characteristics
        Average_Transaction_Duration=(
            "TransactionDuration", "mean"
        ),

        Median_Transaction_Duration=(
            "TransactionDuration", "median"
        ),

        Average_Login_Attempts=(
            "LoginAttempts", "mean"
        ),

        Maximum_Login_Attempts=(
            "LoginAttempts", "max"
        ),

        # Transaction diversity
        Unique_Merchants=(
            "MerchantID", "nunique"
        ),

        Unique_Locations=(
            "Location", "nunique"
        ),

        Unique_Devices=(
            "DeviceID", "nunique"
        ),

        # Temporal activity
        Active_Months=(
            "TransactionMonth", "nunique"
        ),

        First_Transaction_Date=(
            "TransactionDate", "min"
        ),

        Last_Transaction_Date=(
            "TransactionDate", "max"
        )
    )
)


# Calculating transaction span in days

customer_numeric_features["Transaction_Span_Days"] = (
    customer_numeric_features["Last_Transaction_Date"] -
    customer_numeric_features["First_Transaction_Date"]
).dt.total_seconds() / 86400


# Removing raw datetime columns from numerical feature table

customer_numeric_features = (
    customer_numeric_features.drop(
        columns=[
            "First_Transaction_Date",
            "Last_Transaction_Date"
        ]
    )
)


print(
    "Customer-level numerical dataset shape:",
    customer_numeric_features.shape
)

display(customer_numeric_features.head())
```

Proses agregasi berhasil mengubah data transaksi menjadi representasi pada tingkat AccountID. Sebanyak 2.493 transaction records dirangkum menjadi 495 observasi akun pelanggan dengan 22 fitur numerik.

Fitur yang dihasilkan mencakup frekuensi transaksi, karakteristik nominal transaksi, saldo akun, durasi aktivitas, percobaan login, serta variasi merchant, lokasi, dan perangkat.

Hasil ini menjadi fondasi pembentukan customer-level dataset untuk proses clustering.

### 2.11 Categorical Feature Aggregation

Mengubah informasi kategorikal menjadi representasi numerik
berdasarkan frekuensi transaksi masing-masing AccountID.

Atribut yang digunakan:
- TransactionType: Debit, Credit, dan Unknown.
- Channel: ATM, Branch, Online, dan Unknown.

Proses menggunakan frequency aggregation untuk mengetahui
jumlah transaksi berdasarkan jenis dan kanal yang digunakan
setiap akun.

Nilai kosong diperlakukan sebagai kategori Unknown agar
informasi transaksi tetap tercatat tanpa menebak kategori
yang sebenarnya.

Pendekatan ini menghasilkan fitur numerik berbasis frekuensi
sehingga tidak memerlukan LabelEncoder untuk kategori nominal.

```python
# Aggregating transaction type frequencies (Jenis transaksi ; Debit atau Kredit)

transaction_type_data = (
    df_customer_transactions["TransactionType"]
    .where(
        df_customer_transactions["TransactionType"]
        .isin(["Debit", "Credit"]),
        "Unknown"
    )
)

transaction_type_counts = pd.crosstab(
    df_customer_transactions["AccountID"],
    transaction_type_data
)

transaction_type_counts = (
    transaction_type_counts
    .reindex(
        columns=["Debit", "Credit", "Unknown"],
        fill_value=0
    )
)

transaction_type_counts.columns = [
    "Debit_Count",
    "Credit_Count",
    "Unknown_TransactionType_Count"
]

display(transaction_type_counts.head())
```

Atribut TransactionType berhasil diubah menjadi fitur frekuensi berdasarkan jumlah transaksi Debit, Credit, dan Unknown pada setiap AccountID.

Hasil ini memungkinkan karakteristik transaksi masing-masing akun direpresentasikan dalam bentuk numerik tanpa menggunakan LabelEncoder.

Kategori Unknown tetap dipertahankan untuk menunjukkan transaksi yang jenisnya tidak tersedia atau tidak dikenali.

```python
# Aggregating transaction channel frequencies (Kanal Transaksi: Pengelompokkan Perantara Saat Melakukan Transaksi)

channel_data = (
    df_customer_transactions["Channel"]
    .where(
        df_customer_transactions["Channel"]
        .isin(["ATM", "Branch", "Online"]),
        "Unknown"
    )
)

channel_counts = pd.crosstab(
    df_customer_transactions["AccountID"],
    channel_data
)

channel_counts = (
    channel_counts
    .reindex(
        columns=["ATM", "Branch", "Online", "Unknown"],
        fill_value=0
    )
)

channel_counts.columns = [
    "ATM_Count",
    "Branch_Count",
    "Online_Count",
    "Unknown_Channel_Count"
]

display(channel_counts.head())
```

Atribut Channel berhasil diagregasi menjadi jumlah transaksi berdasarkan kanal ATM, Branch, Online, dan Unknown. Fitur ini memungkinkan identifikasi preferensi penggunaan kanal transaksi setiap akun berdasarkan frekuensi aktivitasnya.

Hasil agregasi dapat digunakan untuk menganalisis perbedaan perilaku penggunaan layanan transaksi antar akun.

### 2.12 Combining Features and Creating Behavioral Ratios

Menggabungkan fitur numerik dan kategorikal ke dalam satu
customer-level dataset. Selanjutnya, membentuk fitur proporsi transaksi untuk
merepresentasikan perilaku setiap AccountID.

Fitur rasio yang dibentuk:

1. Debit Ratio
2. Credit Ratio
3. ATM Ratio
4. Branch Ratio
5. Online Ratio

Fitur rasio dihitung berdasarkan jumlah transaksi pada kategori
tersebut dibandingkan dengan keseluruhan transaction records milik akun.

Selain itu, nominal Debit dan Credit yang tidak memiliki transaksi
pada kategori terkait ditetapkan menjadi nol.

Namun, jika transaksi dalam kategori tersebut tersedia tetapi
nominalnya missing, nilai tersebut tetap diperlakukan sebagai
missing value agar tidak disamakan dengan nol.

```python
# Combining numerical and categorical features

customer_df = (
    customer_numeric_features
    .join(transaction_type_counts, how="left")
    .join(channel_counts, how="left")
)


# Setting amount to zero only when the transaction type is absent

customer_df.loc[
    customer_df["Debit_Count"] == 0,
    "Debit_Amount_Sum"
] = 0

customer_df.loc[
    customer_df["Credit_Count"] == 0,
    "Credit_Amount_Sum"
] = 0


# Creating transaction type ratios

customer_df["Debit_Ratio"] = (
    customer_df["Debit_Count"] /
    customer_df["Transaction_Record_Count"]
)

customer_df["Credit_Ratio"] = (
    customer_df["Credit_Count"] /
    customer_df["Transaction_Record_Count"]
)


# Creating channel ratios

customer_df["ATM_Ratio"] = (
    customer_df["ATM_Count"] /
    customer_df["Transaction_Record_Count"]
)

customer_df["Branch_Ratio"] = (
    customer_df["Branch_Count"] /
    customer_df["Transaction_Record_Count"]
)

customer_df["Online_Ratio"] = (
    customer_df["Online_Count"] /
    customer_df["Transaction_Record_Count"]
)


print("Customer-level feature construction completed.")

print("Dataset shape:", customer_df.shape)

display(customer_df.head())
```

Fitur numerik dan kategorikal berhasil digabungkan menjadi satu customer-level dataset dengan 495 observasi dan 34 atribut.

Selain frekuensi transaksi, dibentuk pula fitur rasio Debit, Credit, ATM, Branch, dan Online untuk menggambarkan proporsi aktivitas setiap akun.

Fitur rasio memberikan representasi perilaku yang lebih proporsional sehingga akun dengan jumlah transaksi berbeda tetap dapat dibandingkan berdasarkan pola aktivitasnya.

### 2.15 Handling Missing Values in Customer-Level Features (Menangani Missing Values setelah dilakukan Agregasi)

Menangani missing values yang masih ditemukan pada
candidate feature matrix setelah proses agregasi. Pada tahap transaction-level, missing values tidak diimputasi
secara sembarangan karena dapat mengubah informasi nominal
transaksi yang sebenarnya. Namun, setelah agregasi, beberapa fitur statistik seperti
standard deviation dapat memiliki nilai kosong karena
jumlah observasi yang tidak mencukupi.

Oleh karena itu, **median imputation** diterapkan pada
customer-level feature matrix. Median dipilih karena relatif lebih tahan terhadap distribusi
yang memiliki nilai ekstrem dibandingkan mean.Seluruh nilai yang diimputasi akan dicatat agar perubahan
dataset dapat dipertanggungjawabkan.

Tahap ini dilakukan sebelum standardization. Scaling akan
diterapkan secara terpisah pada tahap eksperimen model.

```python
# Handling missing values in customer-level features

X_customer = X_customer_raw.copy()

# Replacing infinite values if any
X_customer = X_customer.replace(
    [np.inf, -np.inf],
    np.nan
)

# Generating missing value report before imputation
missing_before_imputation = X_customer.isnull().sum()

# Detecting entirely missing features
all_missing_features = (
    X_customer.columns[
        X_customer.isnull().all()
    ].tolist()
)

if all_missing_features:

    raise ValueError(
        f"Features entirely missing: {all_missing_features}. "
        "Review these features before imputation."
    )


# Calculating feature medians
feature_medians = X_customer.median()

# Applying median imputation
X_customer = X_customer.fillna(feature_medians)


# Creating imputation summary
imputation_summary = pd.DataFrame({
    "Feature": X_customer.columns,

    "Missing Before": [
        missing_before_imputation[col]
        for col in X_customer.columns
    ],

    "Imputation Value": [
        feature_medians[col]
        for col in X_customer.columns
    ],

    "Missing After": [
        X_customer[col].isnull().sum()
        for col in X_customer.columns
    ]
})

# Showing only features that required imputation
display(
    imputation_summary[
        imputation_summary["Missing Before"] > 0
    ]
)


print("Missing values after imputation:",
      X_customer.isnull().sum().sum())

print("Final feature matrix shape:",
      X_customer.shape)
```

Ditemukan 57 missing values pada enam fitur hasil agregasi customer-level.

Missing values tersebut ditangani menggunakan median imputation untuk mempertahankan seluruh observasi akun tanpa menghasilkan nilai yang terlalu dipengaruhi oleh distribusi ekstrem.

Setelah imputasi, seluruh fitur kandidat tidak lagi memiliki missing values.
Dengan demikian, feature matrix telah memiliki 495 observasi dan 29 fitur numerik yang lengkap.
